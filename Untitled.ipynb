## Imports

```python
import os
import pandas as pd
import numpy as np
from catboost import CatBoostClassifier
from sklearn.metrics import classification_report
from sklearn.model_selection import TimeSeriesSplit, RandomizedSearchCV
```

## Load Datasets

```python
path = os.getcwd() + "\\data.h5"

X = pd.read_hdf(path, key = "Features")
y = pd.read_hdf(path, key = "Labels")
```

## Keep Chronological Order

```python
order = X["Date"].sort_values().index

X = X.loc[order].reset_index(drop=True)
y = y.loc[order].reset_index(drop=True)
```

## Remove Empty Rows

```python
mask = X.notna().any(axis=1)
X = X.loc[mask].reset_index(drop=True)
y = y.loc[mask].reset_index(drop=True)
```

```python
X.head()
```

```python
categorical_features = [
    "HomeTeam",
    "AwayTeam",
    "Referee",
    "Div"
]
```

## Remove Features that can cause Leakage

```python
drop_columns = [
    "Date",
    "HTR",
    "FTAG",
    "FTHG"
]

X = X.drop(columns=drop_columns)
```

## Split into Training and Testing

```python
split_idx = int(len(X) * 0.8)

X_train = X.iloc[:split_idx].copy()
X_test = X.iloc[split_idx:].copy()

y_train = y.iloc[:split_idx].copy()
y_test = y.iloc[split_idx:].copy()
```

```python
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

## Convert Categorical Features into Strings

```python
for col in categorical_features:
    X_train[col] = X_train[col].fillna("Unknown").astype(str)
    X_test[col] = X_test[col].fillna("Unknown").astype(str)
```

## CatBoost Model

```python
cat_features = [
    X_train.columns.get_loc(col)
    for col in categorical_features
]
```

```python
model = CatBoostClassifier(
    loss_function="MultiClass",
    eval_metric="MultiClass",
    iterations=500,
    learning_rate=0.05,
    depth=6,
    random_seed=42,
    verbose=False
)
```

```python
model.fit(
    X_train,
    y_train,
    cat_features=categorical_features
)
```

```python
y_pred = model.predict(X_test).ravel()
```

```python
print(classification_report(y_test, y_pred))
```

## CatBoost with Hyperparameter Tuning - Raw Data

```python
split_idx = int(len(X) * 0.8)

X_train = X.iloc[:split_idx].copy()
X_test  = X.iloc[split_idx:].copy()

y_train = y.iloc[:split_idx].copy()
y_test  = y.iloc[split_idx:].copy()
```

```python
param_grid = {
    "iterations": [300, 500, 800, 1000, 1200],
    "depth": [4, 5, 6, 7, 8],
    "learning_rate": [0.01, 0.03, 0.05, 0.1],
    "l2_leaf_reg": [1, 3, 5, 10]
}
```

```python
model = CatBoostClassifier(
    loss_function="MultiClass",
    eval_metric="MultiClass",
    random_seed=42,
    verbose=False,
    thread_count=-1
)
```

```python
tscv = TimeSeriesSplit(n_splits=5)

search = RandomizedSearchCV(
    estimator=model,
    param_distributions=param_grid,
    n_iter=30,
    scoring="f1_macro",
    cv=tscv,
    random_state=42,
    n_jobs=-1,
    verbose=2,
    error_score="raise"
)
```

```python
search.fit(
    X_train,
    y_train,
    cat_features=categorical_features
)
```

```python
print(search.best_params_)
print(search.best_score_)
```

```python
best_model = search.best_estimator_
```

```python
y_pred_best_catb = best_model.predict(X_test)
```

```python
print(classification_report(y_test, y_pred_best_catb))
```
