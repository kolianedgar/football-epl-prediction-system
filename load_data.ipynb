```python
import os
import glob
import pandas as pd
```

```python
files = glob.glob(os.getcwd() + "\\data\\*.csv")
```

```python
all_dfs = []

for file in files:
    try:
        temp_df = pd.read_csv(file, encoding='utf-8')
        all_dfs.append(temp_df)
    except Exception as e:
        print(f"❌ Could not read {file} due to: {e}")
```

```python
# Concatenate all successfully read files
if all_dfs:
    df = pd.concat(all_dfs, ignore_index=True)
    print(f"✅ Success! Combined DataFrame shape: {df.shape}")
else:
    print("❌ No files were successfully read.")
```

```python
df = df[df.columns[:list(df.columns).index("B365H")]]

df.columns
```
